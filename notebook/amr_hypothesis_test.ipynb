### Hypothesis 1

This research focuses on a hypothesis test to determine if there is a statistical difference in the amount of terminated phase 3 trials funded by industries and other funders (non-industry).

#### Extracting the variables I need for this project



Needed Variables:
- **Phase**: Phase 3
- **Study status**: Finished = Completed / Terminated
- **Funder Type**: Industry or Other
- **start date**: 2010


 *null* *hypothesis*: There is no difference in the rate of terminated phase 3 trials between industry funded trials and other (non-industry) funded trials.

*alternative* *hypothesis*: There is a difference in the rate of terminated phase 3 trials between industry funded trials andother (non-industry) funded trials.

```python
study_yr = [2010, 2011, 2012, 2013, 2014, 2015, 2016, 2017, 2018, 2019, 2020, 2021, 2022, 2023, 2024, 2025]
funder_study = ['INDUSTRY', 'OTHER']
finished_study = ['COMPLETED', 'TERMINATED']
analysis1_df = ct_df[(ct_df['year'].isin(study_yr)) & (ct_df['Phases'] == 'PHASE3') & (ct_df['Study Status'].isin(finished_study)) & (ct_df['Funder Type'].isin(funder_study))]
```

```python
analysis1_df.sample(12)
```

```python
analysis1_df.info()
```

```python
# Extracting industry to calculate proportion
# Termination rate based on this dataset 2010-2025 = 12.59%

ind = analysis1_df[analysis1_df['Funder Type'] == 'INDUSTRY']

ind_term = ind[ind['Study Status'] == 'TERMINATED']['Study Status'].count()
ind_total = ind[ind['Study Status'].isin(finished_study)]['Study Status'].count()

ind_term_p = ind_term / ind_total

print(ind_term)
print(ind_total)
print(ind_term_p)
```

```python

# Extracting 'other' to calculate proportion
# Termnation rate based on this dataset 2010-2025 = 14.48%

other = analysis1_df[analysis1_df['Funder Type'] == 'OTHER']

other_term = other[other['Study Status'] == 'TERMINATED']['Study Status'].count()
other_total = other[other['Study Status'].isin(finished_study)]['Study Status'].count()

other_term_p = other_term / other_total

print(other_term)
print(other_total)
print(other_term_p)
```

```python
# The 'industry' has a 1.89 percentage point below 'other'. Showing that trials funded by industry have a lesser termination percent than the 'other' funder type

difference = ind_term_p - other_term_p

difference
```

```python
# To calculate if this percentage point is statistically significant or by chance
# running a two sample z-test on proportion
# Standard significance level was chosen as our threshold for this research. if p > 0.05, we fail to reject the null hypothesis. if p< 0.05, we reject the null hypothesis


count = [ind_term, other_term]

nob =[ind_total, other_total]

zstat, pvalue = sm.stats.proportions_ztest(count, nob)

print(zstat)
print(pvalue)
```

p = 0.00069,

p < 0.001,

p < 0.01,

p < 0.05


In this reseach, we reject the null hypothesis which states that there  is equal rate of phase 3 termination betweeen industry and non_industry funders, and conclude that the 1.8 percentage point difference is statisticallly significant, meaning that phase 3 trials by industries has less terminations than 'other'. 





### Hypothesis 2

This research focuses on a hypothesis test to determine if there is a statistical difference in the average size of completed trials between industries and other funders (non-industry).

#### Extracting the variables I need for this project



Needed Variables:
- **Phase**: Phase 3
- **Study status**: Completed
- **Funder Type**: Industry or Other
- **start date**: 2010
- **Enrollment**


 *null* *hypothesis*: There is no difference in the average size of completed phase 3 trials between industry funded trials and other (non-industry) funded trials.
 
*alternative* *hypothesis*: There is a difference in the average size of completed phase 3 trials between industry funded trials andother (non-industry) funded trials.

```python
analysis2_df = analysis1_df[analysis1_df['Study Status'] == 'COMPLETED']
```

```python
# The enrollment variable is has missing values

analysis2_df.info()
```

```python
# Investigating the missing values in the enrollment variable

analysis2_df[analysis2_df['Enrollment'].isna()]
```

```python
# Eleven missing values in the enrollment variable this is very small for a dataset of 14117 rows
# will drop the values.


analysis2_df = analysis2_df.dropna(subset= ['Enrollment'])
```

```python
analysis2_df.info()
```

```python
# CALCULATING THE AVG SIZE OF BOTH GROUPS

ind = analysis2_df[analysis2_df['Funder Type'] == 'INDUSTRY']

other = analysis2_df[analysis2_df['Funder Type'] == 'OTHER']

```

```python
# running a two sample t-test for the mean
# Standard significance level was chosen as our threshold for this research. if p > 0.05, we fail to reject the null hypothesis. if p< 0.05, we reject the null hypothesis


tstat,   pvalue = stats.ttest_ind(ind['Enrollment'], other['Enrollment'], equal_var= False)

print(tstat)
print(pvalue)
```

```python
analysis1_df.to_csv('analysis_df.csv', index= False)
```

t-stat = -1.22 (in respect to Industry trials sizes).

 A negative t-stats in respect to industry trials tells us that trials funded by industries has a lower size than non-industry trial sizes,  a positive t-stats tells us that industry trials have a higher size than non-industry trial sizes, and a ZERO tell us there is no difference between the sizes of trials between industry and non-industry.

p = 0.221

p > 0.05


In this reseach, we fail to reject the null hypothesis which states that there is no difference of phase 3 trial size betweeen industry and non_industry funders, and conclude that our result is not statisticallly significant, meaning there is not enough evidence that phase 3 trials by industries has different average enrollment sizes than 'other'.
